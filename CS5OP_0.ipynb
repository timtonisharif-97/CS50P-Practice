```python
# Indoor Voice
# Ask of user's country
country = input("Where do you come from? ")

# Response in lower letters
country = country.lower()

# Show user's response
print(country)
```

```python
# Playback Speed
# Ask user for favorite food
food = input("What's your favorite food? ")

# replace space in user's response with ...
food = food.replace(" ", "...")

# show user's response
print(food)
```

```python
# Making Faces
# Main function for converting user's response
def main():
  feelings = input("How are you feeling? ")
  answer = convert(feelings)
  print(answer)

# Create a convert function
def convert(answer):
  answer = answer.replace(":)", "😊").replace("):", "😒").replace(" ", "...")

  #return convert function to main function
  return answer

main ()

```

```python
# Tip Calculator
def main():
    dollars = dollars_to_float(input("How much was the meal? "))
    percent = percent_to_float(input("What percentage would you like to tip? "))
    tip = dollars * percent
    print(f"Leave ${tip:.2f}")


def dollars_to_float(d):
    d = float(d.replace("$", ""))
    return d


def percent_to_float(p):
    p = float(p.replace("%", "")) /100
    return p


main()
```
