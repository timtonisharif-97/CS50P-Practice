```python
#   Deep Thought
''' Ask user for the answer to the Great Question of Life,
    the Universe and Everything
'''
answer = input("What's the answer to the Great Question of Life? ").strip().lower()

# Display ‘yes’  if user’s response is 42 or forty-two or forty two
if answer == "42" or answer == "forty-two" or answer == "forty two":
    print("Yes")

# Display ‘No’ for any other response
else:
  print("No")
```

```python
# Home Federal Savings Bank

#Teller’s responding greeting
teller_greeting = input("How did the teller respond? ").strip().lower()

#If greeting is “hello”, customer gets $0

if teller_greeting.startswith("hello"):
  print("$0")

#If greeting starts with “h” but not “hello”, customer gets $20
elif teller_greeting.startswith("h"):
  print("$20")

#If greeting is neither  “hello” nor starts with an “h”
else:
  print("$100")

```

```python
# File Extensions

# Ask user for type of file
file_name = input("Whats your file name? ").strip().lower()

'''File extension based on user’s file name
(.gif,.jpg, .jpeg, .png, .pdf, .txt, .zip)
'''
if file_name.endswith(".gif"):
  print("image/gif")

elif file_name.endswith((".jpg", ".jpeg")):
  print("image/jpeg")

elif file_name.endswith(".png"):
  print("image/png")

elif file_name.endswith(".pdf"):
  print("application/pdf")

elif file_name.endswith(".txt"):
  print("text/plain")

# Respond with  ‘application/octet-stream’
#    if file name is not recognized

else:
  print("application/octet-stream")
```
